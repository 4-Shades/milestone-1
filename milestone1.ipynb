```python
def Interest(principal, rate, time):
    """
    Calculate simple interest.
    
    param principal: The principal amount (initial investment)
    param rate: The annual interest rate (as a percentage)
    param time: The time the money is invested for (in years)
    return: The simple interest earned
    """

    try:
        interest = round((principal * (rate / 100) * time), 2)
    except TypeError:
        print("Error: Please ensure all inputs are numbers.")
        return None

    return interest
```

```python
def totalPayment(interest, principal):
    """
    Calculate the total payment for a loan.

    param interest: The interest amount calculated using the simple interest formula
    param principal: The principal amount (initial investment)
    return: The total payment (principal + interest)
    """
    # Calculate the total payment using simple interest formula
    try:
        total_payment = round((interest + principal), 2)
        return total_payment
    except Exception as e:
        print(f"An error occurred while calculating total payment: {e}")
        return None
```

```python
def main():
    input_amount, input_rate, input_term, isValid = 0, 0, 0, False

    while not isValid:
        try:
            input_amount = float(input("Enter the loan amount: "))
            input_rate = float(input("Enter the interest rate: "))
            input_term = float(input("Enter the loan term (in years): "))
            

            if input_amount <= 0 or input_rate < 0 or input_term <= 0:
                raise ValueError("Loan amount, interest rate, and loan term must be positive numbers.")            
            if input_rate > 100:
                raise ValueError("Interest rate cannot exceed 100%.")
            if type(input_amount) not in [int, float] or type(input_rate) not in [int, float] or type(input_term) not in [int, float]:
                raise ValueError("Invalid input types. Please enter numeric values for loan amount, interest rate, and loan term.")
            

        except ValueError as e:
            print(f"Error: {e}")
            continue

        else:
            isValid = True
            print("Input validation complete.")
            break
    
    
    # Calculate interest payment
    interest = Interest(input_amount, input_rate, input_term)
    
    # Calculate total payment
    total_payment = totalPayment(interest, input_amount)
    
    print(f"The interest for a ₱{input_amount:,.2f} loan with a {input_rate}% interest rate over {input_term} years is ₱{interest:,.2f}.\nThe total payment over the loan term is ₱{total_payment:,.2f}.")
```

```python
main()
assert Interest(100000, 7, 10) == 70000
assert totalPayment(70000, 100000) == 170000

assert Interest(50000, 5, 5) == 12500
assert totalPayment(12500, 50000) == 62500

print("All tests passed successfully.")
```
